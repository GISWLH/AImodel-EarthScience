## Soil: complete fixed baseline and grouped camera audit

This is the same standalone educational pipeline as [Soil Grouped CV Texture Baseline](https://www.kaggle.com/code/giswlh/soil-grouped-cv-texture-baseline), which completed on Kaggle before this combined notebook was built.

Fixed physical DoG features at4.5pixels/mm, sqrt-bin-mass Ridge(alpha1), sample-balanced weights, fold-local scaling, valid monotone output CDFs. No neural checkpoint is needed. Evaluation keeps all photographs of a soil together and reports leave-one-soil-out, proxy-site-out and both phone directions. Proxy groups use nearby sample numbers and are **not verified geographic sites**. This is known-device development evidence, not a blind new-camera trial.

Ideas are credited to YogaNukko's [texture+SAM](https://www.kaggle.com/code/takahitomizunobyts/soil-texture-model-sam-cobble-measurement) and Nomannic's [DINOv2+sqrt-mass Ridge](https://www.kaggle.com/code/nomannic19/cv-psgsdi-dinov2-features-sqrt-mass-ridge). No third-party neural weights or competition data are redistributed.

```python
"""Train-only, standalone physical texture baseline and grouped domain audit.
Original educational implementation by giswlh, AI-assisted, 2026-10-19.
"""
import os
from pathlib import Path
import re
import unicodedata
import json
import platform
import hashlib
import time
import numpy as np
import pandas as pd
import scipy
from scipy.ndimage import gaussian_filter
from sklearn.linear_model import Ridge
from sklearn.preprocessing import StandardScaler
import sklearn
from PIL import Image, ImageOps
import PIL

START=time.monotonic()
DIAMETERS=np.array([.002,.0063,.02,.063,.2,.63,2,6.3,20,63,200])
COLUMNS=['0.002','0.0063','0.02','0.063','0.2','0.63','2','6.3','20','63','200']
TARGET_PPM=4.5
ALPHA=1.0

def canonical(text):
    text=str(text).translate(str.maketrans({'ü':'ue','ö':'oe','ä':'ae','ß':'ss'}))
    return re.sub('[^a-z0-9]','',unicodedata.normalize('NFKD',text).encode('ascii','ignore').decode().lower())

def emd(truth,prediction):
    return np.sum(np.abs(np.asarray(truth)[:,:-1]-np.asarray(prediction)[:,:-1])*np.diff(np.log10(DIAMETERS)),axis=1)

def check_cdf(values):
    a=np.asarray(values)
    assert a.ndim==2 and a.shape[1]==11 and np.isfinite(a).all()
    assert a.min()>=0 and a.max()<=100 and (np.diff(a,axis=1)>=-1e-10).all()
    assert np.all(a[:,-1]==100.)

def extract_features(root,labels):
    ppm=pd.read_csv(root/'ppm_updated.csv').set_index('phone')
    ids=list(labels.index);phones=sorted(ppm.index,key=lambda p:len(canonical(p)),reverse=True);records=[]
    # Only Training directories are opened. No test photographs or hidden labels.
    for path in sorted(root.rglob('*')):
        if not path.is_file() or path.suffix.lower() not in {'.jpg','.jpeg'}:continue
        if not any('training' in part.lower() for part in path.relative_to(root).parts):continue
        key=canonical(path.stem);matches=[sid for sid in ids if canonical(sid) in key]
        phone=next((p for p in phones if key.startswith(canonical(p))),None)
        if len(matches)!=1 or phone is None:raise ValueError('Ambiguous training photograph mapping: '+path.name)
        with Image.open(path) as im:
            # Scale from the original file dimensions before EXIF orientation.
            effective_ppm=float(ppm.loc[phone,'ppm'])*max(im.size)/max(ppm.loc[phone,'width'],ppm.loc[phone,'height'])
            image=ImageOps.exif_transpose(im).convert('L')
            scale=TARGET_PPM/effective_ppm
            image=image.resize((max(8,round(image.width*scale)),max(8,round(image.height*scale))),Image.Resampling.LANCZOS)
            gray=np.asarray(image,dtype=np.float32)
        height,width=gray.shape;gray=gray[height//4:3*height//4,width//4:3*width//4]
        contrast=float(gray.std())+1e-6
        smooth=[gaussian_filter(gray,sigma=mm*TARGET_PPM) for mm in [.25,.5,1,2,4,8]]
        features=[float(np.log(np.std(a-b)/contrast+1e-6)) for a,b in zip(smooth[:-1],smooth[1:])]
        records.append({'sample_id':matches[0],'phone':phone,'file':path.name,**{f'dog_{i}':v for i,v in enumerate(features)}})
    frame=pd.DataFrame(records)
    assert set(frame.sample_id)==set(ids),'Missing training photo coverage'
    assert np.isfinite(frame.filter(like='dog_').to_numpy()).all()
    return frame

def proxy_sites(ids):
    # Development grouping heuristic only: these are NOT verified geographic sites.
    group=0;previous=None;result={}
    for sid in sorted(ids,key=lambda s:(s[0],int(s[1:]))):
        if previous is None or sid[0]!=previous[0] or int(sid[1:])-int(previous[1:])>12:group+=1
        result[sid]=group;previous=sid
    return result

def predict(fit,valid,labels):
    assert not set(fit.sample_id)&set(valid.sample_id),'Soil sample leakage'
    cols=[f'dog_{i}' for i in range(5)]
    # All fitted statistics and model weights are fold-local; each soil has total weight1.
    weights=1/fit.groupby('sample_id').sample_id.transform('size').to_numpy()
    scaler=StandardScaler().fit(fit[cols],sample_weight=weights)
    x=scaler.transform(fit[cols]);v=scaler.transform(valid[cols])
    y=labels.loc[fit.sample_id].to_numpy()/100
    mass=np.diff(np.c_[np.zeros(len(y)),y],axis=1)
    assert mass.min()>=-1e-10
    model=Ridge(alpha=ALPHA).fit(x,np.sqrt(np.maximum(mass,0)),sample_weight=weights)
    root_mass=np.maximum(model.predict(v),0);mass=root_mass**2
    zero=mass.sum(axis=1)<=1e-12;mass[zero]=1.
    mass/=mass.sum(axis=1,keepdims=True)
    curve=np.cumsum(mass,axis=1)*100;curve=np.clip(curve,0,100);curve[:,-1]=100
    result=pd.DataFrame(curve,index=valid.sample_id,columns=COLUMNS).groupby(level=0).mean()
    check_cdf(result);return result

def evaluate(photos,labels,sites):
    protocols={
        'leave_one_soil_out':{'groups':{s:s for s in labels.index},'train_phone':None,'valid_phone':None},
        'leave_proxy_site_out':{'groups':sites,'train_phone':None,'valid_phone':None},
        'Motorola_to_Samsung':{'groups':sites,'train_phone':'Motorola Edge','valid_phone':'Samsung A52'},
        'Samsung_to_Motorola':{'groups':sites,'train_phone':'Samsung A52','valid_phone':'Motorola Edge'},
    }
    summaries=[];oof_rows=[]
    for name,cfg in protocols.items():
        available=photos if cfg['valid_phone'] is None else photos[photos.phone==cfg['valid_phone']]
        predicted=[]
        for group in sorted(set(cfg['groups'].values()),key=str):
            held={s for s,g in cfg['groups'].items() if g==group}
            fit=photos[~photos.sample_id.isin(held)];valid=available[available.sample_id.isin(held)]
            if cfg['train_phone'] is not None:fit=fit[fit.phone==cfg['train_phone']]
            if valid.empty:continue
            assert len(set(fit.sample_id))>=2 and not set(fit.sample_id)&held
            p=predict(fit,valid,labels);error=emd(labels.loc[p.index].to_numpy(),p.to_numpy())
            for sid,e,curve in zip(p.index,error,p.to_numpy()):
                predicted.append(float(e));oof_rows.append({'protocol':name,'sample_id':sid,'held_group':group,'emd':float(e),**dict(zip(COLUMNS,curve.tolist()))})
        assert len(predicted)==available.sample_id.nunique()
        summaries.append({'protocol':name,'n_soils':len(predicted),'mean_emd':float(np.mean(predicted)),'worst_soil_emd':float(max(predicted))})
    return summaries,pd.DataFrame(oof_rows)

def run_soil_training_audit():
    override=os.environ.get('SOIL_DATA_ROOT')
    if override:root=Path(override)
    else:
        files=list(Path('/kaggle/input').rglob('Training_labels_updated.csv'))
        if len(files)!=1:raise RuntimeError('Attach soil-grain-size-from-photos competition data; expected one updated training label file.')
        root=files[0].parent
    out=Path(os.environ.get('SOIL_OUTPUT_ROOT','/kaggle/working'));out.mkdir(parents=True,exist_ok=True)
    labels=pd.read_csv(root/'Training_labels_updated.csv',dtype={'sample_id':str}).set_index('sample_id')
    labels.columns=[format(float(c),'g') for c in labels.columns]
    labels=labels[COLUMNS];check_cdf(labels)
    photos=extract_features(root,labels);sites=proxy_sites(labels.index)
    summary,oof=evaluate(photos,labels,sites)
    report={'method':'fixed physical DoG + sqrt-mass Ridge','hyperparameters':{'target_ppm':TARGET_PPM,'alpha':ALPHA,'gaussian_scales_mm':[.25,.5,1,2,4,8]},'n_training_soils':len(labels),'n_training_photos':len(photos),'n_proxy_groups':len(set(sites.values())),'results':summary,'formal_submission':False,'leaderboard_score':None,'test_photos_read':False,'labels_sha256':hashlib.file_digest((root/'Training_labels_updated.csv').open('rb'),'sha256').hexdigest(),'environment':{'python':platform.python_version(),'numpy':np.__version__,'pandas':pd.__version__,'scipy':scipy.__version__,'sklearn':sklearn.__version__,'Pillow':PIL.__version__},'elapsed_seconds':time.monotonic()-START}
    (out/'validation_report.json').write_text(json.dumps(report,indent=2,allow_nan=False)+'\n');oof.to_csv(out/'grouped_oof.csv',index=False)
    print(pd.DataFrame(summary).to_string(index=False));print(json.dumps({k:report[k] for k in ['n_training_soils','n_training_photos','n_proxy_groups','elapsed_seconds','test_photos_read','formal_submission']},indent=2))
    return report,oof

```

```python
"""Small, explicit training cores for a three-task research notebook.

These educational kernels do not reproduce the complete best-score pipelines.
Only real historical/labelled training data are accepted. No automatic submissions.
"""
from pathlib import Path
import numpy as np
import pandas as pd

CARBON_AGES=[1,10,20,30,50,70,100,150,200,250,300,350,400,450,500]
CARBON_SCALE=np.array([10.90645,3.39567],dtype=np.float32)

def carbon_forcing_features(x):
    """Actual (site,40years,12months,features) -> reference + seasonal summaries."""
    assert x.ndim==4 and x.shape[1:3]==(40,12) and np.isfinite(x).all()
    annual=x.mean(axis=2,dtype=np.float32);centered_year=np.arange(40,dtype=np.float32)-19.5
    early=annual[:,:5].mean(1);late=annual[:,-5:].mean(1)
    reference=np.concatenate([annual.mean(1),annual.std(1),(annual*centered_year[None,:,None]).sum(1)/(centered_year@centered_year),late,*[annual[:,i:i+10].mean(1) for i in range(0,40,10)],np.clip((late-early)/(np.abs(early)+1e-6),-10,10)],axis=1)
    seasonal=x.mean(1)
    return reference.astype(np.float32),np.concatenate([seasonal.std(1),seasonal.min(1),seasonal.max(1),late-early],axis=1).astype(np.float32)

def carbon_site_split(site_ids,seed=42,validation_fraction=.2):
    ids=np.asarray(site_ids);assert len(np.unique(ids))==len(ids) and len(ids)>=10
    rng=np.random.default_rng(seed);order=rng.permutation(ids);n=max(2,int(len(ids)*validation_fraction))
    train,valid=np.sort(order[n:]),np.sort(order[:n]);assert not np.intersect1d(train,valid).size
    return train,valid

def carbon_training_core(data_root,max_sites=512,seed=42,rounds=200):
    """Fixed LightGBM final-year baseline; all15ages of a site have one split.

    For bounded educational execution use the first max_sites training rows.
    This subset is not representative global validation and must not be used as
    evidence of an improved competitive model. No test inputs are opened.
    """
    from lightgbm import LGBMRegressor
    root=Path(data_root)/'data_global';x=np.load(root/'glob_X_fea.npy',mmap_mode='r')
    n=min(max_sites,len(x));assert n>=10
    reference,seasonal=carbon_forcing_features(np.asarray(x[:n]))
    initial=[];target=[]
    for age in CARBON_AGES:
        y=np.load(root/f'glob_Y{age:03d}.npy',mmap_mode='r');assert y.shape[1]==41
        initial.append(np.asarray(y[:n,0,-1,:],dtype=np.float32));target.append(np.asarray(y[:n,40,-1,:2],dtype=np.float32))
    initial=np.stack(initial,1);target=np.stack(target,1);assert initial.shape==(n,15,7)
    ages=np.asarray(CARBON_AGES,dtype=np.float32)[None,:,None]
    rows=np.concatenate([np.repeat(reference,15,axis=0),initial.reshape(-1,7),(initial/ages).reshape(-1,7),np.tile(CARBON_AGES,n)[:,None]],axis=1).astype(np.float32)
    train,valid=carbon_site_split(np.arange(n),seed)
    train_rows=np.concatenate([np.arange(i*15,(i+1)*15) for i in train]);valid_rows=np.concatenate([np.arange(i*15,(i+1)*15) for i in valid])
    truth=target.reshape(-1,2)/CARBON_SCALE;prediction=np.empty((len(valid_rows),2))
    for k in range(2):
        model=LGBMRegressor(n_estimators=rounds,num_leaves=31,learning_rate=.05,min_child_samples=30,reg_alpha=.5,reg_lambda=1.,random_state=seed,n_jobs=2,verbosity=-1)
        model.fit(rows[train_rows],truth[train_rows,k]);prediction[:,k]=np.maximum(model.predict(rows[valid_rows]),0)
    assert np.isfinite(prediction).all()
    return {'kind':'bounded_real_training_core_execution','representative_validation':False,'official_score':False,'test_inputs_read':False,'training_sites':len(train),'held_sites':len(valid),'all_ages_grouped':True,'seed':seed,'rounds':rounds,'target_year':40,'output_order':['height','agb'],'official_unit_divisors':CARBON_SCALE.tolist(),'normalized_held_MSE':float(np.mean((prediction-truth[valid_rows])**2))}

```

## ClimX: forcing-conditioned daily distributions before indices

The existing submitted pipeline uses fitted forcing memory, two aerosol PCs, monthly conditional quantiles, coherent historical weather templates and a verified native-grid index implementation. Its development run used32weather members and the complete submission128members. All7daily variables are generated before the15official indices. Frozen historical thresholds and evaluator/backend parity matter.

Below is an intentionally small, original training core: fit monthly quantiles against CO2 design rows on3real historical years and map one shared training weather year into a fourth held year at4cells. It generates all7daily fields and checks finite values, positive precipitation and temperature ordering. This illustrates the mechanism; it omits the full forcing/PCA model, historical seasonal demeaning, production tail coordinates, full weather ensemble and official index calculation. **Its daily MAEs are implementation diagnostics, not official rnNSE or full model validation.** CPU is default; `torch_cuda` can be explicitly selected in an environment with authorized GPU access.

Source idea: [maverickss26/climx-sota-main-track](https://www.kaggle.com/code/maverickss26/climx-sota-main-track). Historical raw data are opened read-only. The future SSP245 target file is never read here.

```python
CLIMX_VARIABLES=['tas','tasmax','tasmin','pr','huss','psl','sfcWind']
CLIMX_PROBS=np.array([.01,.02,.035,.05,.075,.1,.15,.2,.25,.3,.35,.4,.45,.5,.55,.6,.65,.7,.75,.8,.85,.9,.925,.95,.965,.98,.99])

def climx_month_windows():
    centers=np.round((np.arange(12)+.5)*365/12-.5).astype(int)
    return (centers[:,None]+np.arange(-30,31)[None,:])%365

def climx_fit_quantile_core(design,daily,ridge=None,backend='numpy'):
    """Daily arrays(year,365,cell). Fit each month/quantile/cell on forcing.

    numpy is the default. torch_cuda explicitly uses an already-authorized GPU.
    This core omits the full pipeline's fitted forcing memory/aerosol PCA,
    deseasonalization, calibrated tails, weather ensemble and official indices.
    """
    design=np.asarray(design,dtype=np.float64);daily=np.asarray(daily,dtype=np.float32)
    assert daily.shape[:2]==(len(design),365) and np.isfinite(daily).all()
    windows=climx_month_windows()
    q=np.quantile(daily[:,windows,:],CLIMX_PROBS,axis=2).transpose(1,2,0,3)
    penalties=np.r_[0.,np.ones(design.shape[1]-1)] if ridge is None else np.asarray(ridge)
    a=design.T@design+np.diag(penalties);b=design.T@q.reshape(len(design),-1)
    if backend=='torch_cuda':
        import torch
        assert torch.cuda.is_available()
        with torch.inference_mode():coef=torch.linalg.solve(torch.as_tensor(a,device='cuda'),torch.as_tensor(b,device='cuda')).cpu().numpy()
    else:
        assert backend=='numpy';coef=np.linalg.solve(a,b)
    return np.ascontiguousarray(coef.reshape(design.shape[1],12,len(CLIMX_PROBS),daily.shape[-1]))

def climx_daily_knots(coefficients,x,precipitation=False):
    monthly=np.sort(np.einsum('p,pmkc->mkc',np.asarray(x),coefficients,optimize=False),axis=1)
    if precipitation:monthly=np.maximum(monthly,0)
    position=(np.arange(365)+.5)*12/365-.5;left=np.floor(position).astype(int);w=(position-left)[:,None,None]
    return monthly[left%12]*(1-w)+monthly[(left+1)%12]*w

def climx_map_weather(template,source_knots,target_knots):
    """Same real template days/cells across7variables preserve joint day pairing.

    This bounded reference kernel interpolates on fixed quantile probabilities;
    it clips probabilities to[0,1] and is not the production tail extrapolator.
    """
    output=np.empty_like(template,dtype=np.float32)
    for d in range(len(template)):
        for c in range(template.shape[1]):
            p=np.interp(template[d,c],source_knots[d,:,c],CLIMX_PROBS)
            output[d,c]=np.interp(p,CLIMX_PROBS,target_knots[d,:,c])
    return output

def climx_historical_core(data_root,backend='numpy'):
    """Real4year/4cell mechanics check; year4held from fitting. No rnNSE claim."""
    import xarray as xr
    files=list(Path(data_root).rglob('*hist_train_1850_1931.nc'))
    assert len(files)==1,'Attach competition historical lite NetCDF data'
    with xr.open_dataset(files[0],decode_times=False) as ds:
        assert ds.time.attrs['calendar'] in ('noleap','365_day')
        assert np.array_equal(ds.time.values[:4*365],np.arange(4*365)), 'Expected complete daily years'
        co2=np.asarray(ds.CO2_LBC.isel(forcing_scenario=0).values[:4],dtype=np.float64)
        center=co2[:3].mean();scale=max(float(co2[:3].std()),1e-12);design=np.c_[np.ones(4),(co2-center)/scale]
        raw={}
        for v in CLIMX_VARIABLES:
            arr=np.asarray(ds[v].isel(forcing_scenario=0,time=slice(0,4*365),lat=6,lon=slice(0,4)).values,dtype=np.float32)
            if v=='pr':
                assert ds[v].attrs.get('units')=='kg m-2 s-1';arr*=86400.
            raw[v]=arr.reshape(4,365,4)
    generated={};diagnostics=[]
    for v,arr in raw.items():
        transformed=np.cbrt(np.maximum(arr,0)) if v=='pr' else arr
        coef=climx_fit_quantile_core(design[:3],transformed[:3],backend=backend)
        # The template is year3, always inside the training partition.
        source=climx_daily_knots(coef,design[2],v=='pr');target=climx_daily_knots(coef,design[3],v=='pr')
        z=climx_map_weather(transformed[2],source,target)
        generated[v]=np.maximum(z,0)**3 if v=='pr' else z
        if v in ['pr','huss','sfcWind']:generated[v]=np.maximum(generated[v],0)
    generated['tasmax']=np.maximum(generated['tasmax'],generated['tas']);generated['tasmin']=np.minimum(generated['tasmin'],generated['tas'])
    for v,p in generated.items():
        assert np.isfinite(p).all()
        diagnostics.append({'variable':v,'historical_hold_year':1853,'daily_MAE_in_variable_units':float(np.mean(abs(p-raw[v][3]))),'minimum_generated':float(p.min()),'maximum_generated':float(p.max())})
    return {'kind':'limited_real_historical_mechanics_check','official_rnNSE':None,'representative_validation':False,'test_targets_read':False,'train_years':[1850,1851,1852],'held_year':1853,'cells':4,'backend':backend,'all7daily_fields_generated':True,'diagnostics':diagnostics}
```
